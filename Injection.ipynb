```python
import pandas as pd
import numpy as np

# Load filtered base data
df = pd.read_csv('Dataset.csv')
df_filtered = df[df['type'].isin(['TRANSFER', 'CASH_OUT'])].copy()

def inject_anomalies(df, sparsity_level):
    """
    Injects missing values (NaNs) and structural noise across financial features
    at controlled rates: 0.05 (5%), 0.10 (10%), or 0.20 (20%).
    """
    df_corrupted = df.copy()
    target_cols = ['amount', 'oldbalanceOrg', 'newbalanceOrig', 'oldbalanceDest', 'newbalanceDest']
    
    # Calculate number of cells to mask based on sparsity level
    for col in target_cols:
        mask = np.random.rand(len(df_corrupted)) < sparsity_level
        df_corrupted.loc[mask, col] = np.nan
        
    return df_corrupted

# Generate the 3 noisy datasets
df_noise_5 = inject_anomalies(df_filtered, 0.05)
df_noise_10 = inject_anomalies(df_filtered, 0.10)
df_noise_20 = inject_anomalies(df_filtered, 0.20)

# Save corrupted versions for validation and repair testing
df_noise_5.to_csv('paysim_noise_5.csv', index=False)
df_noise_10.to_csv('paysim_noise_10.csv', index=False)
df_noise_20.to_csv('paysim_noise_20.csv', index=False)

print("✅ Successfully generated 5%, 10%, and 20% corrupted dataset baselines!")
```
